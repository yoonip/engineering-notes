# RC step response
저항과 커패시터로 구성된 1차 회로의 충전 응답을 계산한다.

입력이 일정하고 초기 전압이 0일 때 $v(t)=V_{in}(1-e^{-t/(RC)})$다. R=1 kΩ, C=1 µF이면 시정수는 1 ms다.

```python
from math import exp

R, C, vin = 1000.0, 1e-6, 3.3
dt, steps = 1e-5, 500
v = 0.0
for _ in range(steps):
    v += dt * (vin - v) / (R * C)
t = dt * steps
exact = vin * (1.0 - exp(-t / (R * C)))
print(f"Euler={v:.6f}, analytic={exact:.6f}, error={abs(v-exact):.6f}")
```

## 해석해 그래프
횡축은 ms, 종축은 V다. 입력은 3.3 V이며 시간에 따라 그 값에 접근한다.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0.0, 0.01, 1001)
v = 3.3 * (1.0 - np.exp(-t / 0.001))
plt.plot(t * 1000, v)
plt.xlabel("Time [ms]")
plt.ylabel("Voltage [V]")
plt.grid(True)
plt.show()
```
